```python
#plotting good and bad runs for comparison
#bad runs with long seizure window: HUP111 run01, HUP181 run02/run03, HUP173 run01
#bad runs with short seizure window: HUP074 run04/05
#good runs with long seizure window: HUP074 run01, HUP146 run01, HUP138 run02

from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# Add your NPZ files here
NPZ_FILES = [
    #good runs
    Path(r"D:\HUP_processed\pipeline_b_car_0p5_120_128\sub-HUP074\sub-HUP074_ses-presurgery_task-ictal_acq-ecog_run-01.npz"),
    Path(r"D:\HUP_processed\pipeline_b_car_0p5_120_128\sub-HUP146\sub-HUP146_ses-presurgery_task-ictal_acq-seeg_run-01.npz"),
    Path(r"D:\HUP_processed\pipeline_b_car_0p5_120_128\sub-HUP138\sub-HUP138_ses-presurgery_task-ictal_acq-seeg_run-02.npz"),
    #bad runs
    Path(r"D:\HUP_processed\pipeline_b_car_0p5_120_128\sub-HUP111\sub-HUP111_ses-presurgery_task-ictal_acq-ecog_run-01.npz"),
    Path(r"D:\HUP_processed\pipeline_b_car_0p5_120_128\sub-HUP181\sub-HUP181_ses-presurgery_task-ictal_acq-seeg_run-02.npz"),
    Path(r"D:\HUP_processed\pipeline_b_car_0p5_120_128\sub-HUP173\sub-HUP173_ses-presurgery_task-ictal_acq-seeg_run-01.npz"),
]

def print_basic_info(name, data):
    X = data["X"]
    y = data["y"]
    y_train = data["y_train"]
    t_bounds = data["t_bounds"]
    rms_z = data["rms_z"]
    peak_z = data["peak_z"]

    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)

    print(f"X shape       : {X.shape}, dtype={X.dtype}")
    print(f"y shape       : {y.shape}, dtype={y.dtype}")
    print(f"y_train shape : {y_train.shape}, dtype={y_train.dtype}")
    print(f"t_bounds shape: {t_bounds.shape}, dtype={t_bounds.dtype}")
    print(f"rms_z shape   : {rms_z.shape}, dtype={rms_z.dtype}")
    print(f"peak_z shape  : {peak_z.shape}, dtype={peak_z.dtype}")

    n_total = len(y)
    n_ictal = int((y == 1).sum())
    n_nonictal = int((y == 0).sum())
    n_ignore = int((y_train == -1).sum())
    n_train_ictal = int((y_train == 1).sum())
    n_train_nonictal = int((y_train == 0).sum())

    print("\nClass counts")
    print(f"  total windows     : {n_total}")
    print(f"  ictal (y==1)      : {n_ictal}")
    print(f"  non-ictal (y==0)  : {n_nonictal}")
    print(f"  ignored (y_train) : {n_ignore}")
    print(f"  train ictal       : {n_train_ictal}")
    print(f"  train non-ictal   : {n_train_nonictal}")

    def cls_stats(arr, mask, label):
        if mask.sum() == 0:
            print(f"  {label:<12}: no samples")
            return
        vals = arr[mask]
        print(
            f"  {label:<12}: n={mask.sum():4d}, "
            f"mean={vals.mean():8.4f}, std={vals.std():8.4f}, "
            f"min={vals.min():8.4f}, max={vals.max():8.4f}"
        )

    print("\nrms_z by class")
    cls_stats(rms_z, y == 0, "non-ictal")
    cls_stats(rms_z, y == 1, "ictal")

    print("\npeak_z by class")
    cls_stats(peak_z, y == 0, "non-ictal")
    cls_stats(peak_z, y == 1, "ictal")

    pos_windows = t_bounds[y == 1]
    if len(pos_windows) > 0:
        print("\nIctal window span")
        print(f"  first ictal window: {pos_windows[0]}")
        print(f"  last ictal window : {pos_windows[-1]}")

def plot_npz_comparison(name, data):
    y = data["y"]
    y_train = data["y_train"]
    t_bounds = data["t_bounds"]
    rms_z = data["rms_z"]
    peak_z = data["peak_z"]

    # Use window center time for plotting
    t_center = t_bounds.mean(axis=1)

    fig, ax = plt.subplots(figsize=(14, 5))

    ax.plot(t_center, rms_z, label="rms_z")
    ax.plot(t_center, peak_z, label="peak_z")

    # Plot true labels and train labels on secondary axis
    ax2 = ax.twinx()
    ax2.step(t_center, y, where="mid", label="y (test label)")
    ax2.step(t_center, y_train, where="mid", label="y_train (-1 ignore, 0 non-ictal, 1 ictal)")

    ax.set_title(name)
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("rms_z / peak_z")
    ax2.set_ylabel("labels")

    lines1, labels1 = ax.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax.legend(lines1 + lines2, labels1 + labels2, loc="upper right")

    plt.tight_layout()
    plt.show()

for npz_path in NPZ_FILES:
    data = np.load(npz_path)
    print_basic_info(npz_path.name, data)
    plot_npz_comparison(npz_path.name, data)
```
